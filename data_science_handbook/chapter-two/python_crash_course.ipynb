### Sets

Sets are used to store not-duplicate values only.

```python
one_to_ten = [n for n in range(1, 11)]
one_to_ten
```

```python
one_to_ten_duplicate = [n for n in range(1, 11)]
one_to_ten_duplicate
```

```python
one_to_ten.extend(one_to_ten_duplicate)
one_to_ten
```

```python
set_one_to_ten = set(one_to_ten)
set_one_to_ten
```

It removed the duplicate numbers and only the not duplicated remained. This helps our use of the operator `in`, since the operator looks for all values stored in a list, tuple or dict.

The best way to use this operator to make faster lookups is converting a list into a set, and then using the operator.

### Assertions

Assertions are used to assure that the code is working as it should. Use for testing function-behavior or code in general.

```python
import random

random.shuffle(one_to_ten)

def smallest_item(x_list):
    return min(x_list)

assert smallest_item(one_to_ten) == 1
assert smallest_item([10, 20, 5, 40]) == 5, "smallest item of this list should be 5"

print(smallest_item(one_to_ten))
print(smallest_item([10, 20, 5, 40]))
```

### OOP fundamentals

```python
class CountingClicker():
    """classes should have doctrings"""

    def __init__(self, count = 0):
        self.count = count

    def __repr__(self):
        return f"CountingClicker(count={self.count})"

    def click(self, num_times = 1):
        self.count += num_times

    def read(self):
        return self.count

    def reset(self):
        self.count = 0
```

```python
clicker = CountingClicker()
assert clicker.read() == 0, "clicker should start with count 0"
```

```python
clicker.click()
clicker.click()
assert clicker.read() == 2, "after two clicks, clicker should have count 2"
```

```python
clicker.reset()
assert clicker.read() == 0, "after reset, clicker should be back to 0"
```

```python
class NoResetClicker(CountingClicker):
    """this class contains the same methods as the CountingClicker class"""
    def reset(self):
        pass
```

```python
diff_clicker = NoResetClicker()
assert diff_clicker.read() == 0
```

```python
diff_clicker.click()
assert diff_clicker.read() == 1
```

```python
diff_clicker.reset()
assert diff_clicker.read() == 1, "clicker count should be the same because reset() doesnt do anything"
```
