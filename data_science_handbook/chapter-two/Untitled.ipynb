### First class functions

In Python we can attribute functions to variable. For me it's a bit confusing, but it's quise useful as well.

```python
def double(x):
    return x*2

def apply_to_one(f):
    return f(1)

my_double = double #double() becomes a variable
x = apply_to_one(my_double)

x
```

### Lambda functions

Python makes use of lambda functions, where we can make very little (but powerful) functions to implement in our code.

```python
y = apply_to_one(lambda x:x+4)
y
```

### Controlling the default-behavior of a function

We can make functions return some behaviour default by just specifying on the function declaration. This way, if a fuction is called, and no arguments are passed, it just return what we specified.

```python
def name(first="What's his name", last="Joe Schmo"):
    return first + " " + last

name("Pedro", "Lopes")
```

```python
name("Pedro")
```

```python
name(last="Lopes")
```

### String formats

### Exceptions

Exceptions are used for handling program errors.

```python
try:
    print(0/0)
except ZeroDivisionError:
    print("cannot divide by zero")
```

### List handling

#### Slices and strides

We can handle the range of lists slicing its size to fit our criteria.

```python
first_three = x[:3]
first_three
```

```python
three_to_end = x[3:]
three_to_end
```

```python
one_to_four = x[1:5]
one_to_four
```

```python
last_three = x[-3:]
last_three
```
