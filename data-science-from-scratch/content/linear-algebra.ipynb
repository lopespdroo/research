# Linear Algebra

We can visualize this chapter as going under the hood of how the library `numpy` works*. So, here it's covered vector addition, multiplication, and some other operations with vectors.

*`numpy` doesn't quite works this way, but since it's adopted by everyone to manipulate arrays, this is a good analogy to make.

In normal Python we don't have a default vector type, so we need to use a type alias to make the process of creating a vector easier.

```python
from typing import List

Vector = List[float]
```

## Vector addition

```python
def add(v: Vector, w: Vector) -> Vector:
    """Sum corresponding elements"""
    assert len(v) == len(w), "vectors must be the same length"
    return [v_i + w_i for v_i, w_i in zip(v, w)]

assert add([1, 2, 3], [4, 5, 6]) == [5, 7, 9]
```

## Vector substraction

```python
def subtract(v: Vector, w: Vector) -> Vector:
    """Substract corresponding elements"""
    assert len(v) == len(w), "vectors must be the same length"
    return [v_i - w_i for v_i, w_i in zip(v, w)]

assert subtract([4, 5, 6], [1, 2, 3]) == [3, 3, 3]
```

## Sum of vectors

```python
def vector_sum(vectors: List[Vector]) -> Vector:
    """Sum all vector corresponding elements"""
    assert vectors, "no vectors provided!"

    num_elements = len(vectors[0])
    assert all(num_elements == len(v) for v in vectors), "vectors should have the same size!"

    return [sum(vector[i] for vector in vectors)
            for i in range(num_elements)]

assert vector_sum([[1, 2], [3, 4], [5, 6], [7, 8]]) == [16,20]
```

## Scalar multiplication

```python
def scalar_mult(v: Vector, n: float) -> Vector:
    return [v_i * n for v_i in v]

scalar_mult([1, 2], 3)
```

## Mean of a vector

```python
def vector_mean(vectors: List[Vector]) -> Vector:
    """Computes the average of the elements"""
    assert vectors, "no vectors provided"

    num_elements = len(vectors[0])
    assert all(num_elements == len(v) for v in vectors), "vectors should be the same size!"

    n = len(vectors)

    return scalar_mult(vector_sum(vectors), 1/n)

assert vector_mean([[1,2], [3, 4], [5, 6]]) == [3, 4]
```

```python
def dot(v: Vector, w: Vector) -> float:
    """Computes the sum of the product of the elements"""
    assert len(v) == len(w), "vectors should be the same size!"
    
    return sum([v_i * w_i for v_i, w_i in zip(v, w)])

assert dot([1, 2, 3], [4, 5, 6]) == 32
```

```python
def sum_of_squares(v: Vector) -> float:
    """Computes the sum of each element squared"""
    assert v, "no vector provided"

    return dot(v, v)

assert sum_of_squares([1, 2, 3]) == 14
```

```python
import math

def magnitude(v: Vector) -> float:
    """Returns the magnitude of the vector"""
    return math.sqrt(sum_of_squares(v))

assert magnitude([3, 4]) == 5
```

```python
def distance(v: Vector, w: Vector) -> float:
    return magnitude(subtract(v, w))
```

```python
from typing import Tuple

def shape(A: Matrix) -> Tuple[int, int]:
    """Returns the numbers of lines and columns of A"""
    num_rows = len(A)
    num_cols = len(A[0])

    return num_rows, num_cols

assert shape([[1, 2, 3], [4, 5, 6]]) == (2, 3)
```

```python
def get_row(A: Matrix, i: int) -> Vector:
    """Return the i-th line of A (as a vector)"""
    return A[i]

def get_column(A: Matrix, j: int) -> Vector:
    return [A[i][j] for A_i in A]
```

```python
from typing import Callable

def make_matrix(num_rows: int,
                num_cols: int,
                entry_fn: Callable[[int, int], float]) -> Matrix:
    """
    Returns a matrix A: num_rows x num_cols
    where its entry (i, j) is entry_fn(i, j)
    """
    return [[entry_fn(i, j)
             for j in range(num_cols)]
             for i in range(num_rows)]
```

```python
def identity_matrix(n: int) -> Matrix:
    return make_matrix(n, n, lambda i, j: 1 if i == j else 0)
```
